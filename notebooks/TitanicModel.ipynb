```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

train = pd.read_csv("../data/train.csv")
test = pd.read_csv("../data/test.csv")
```

```python
num_cols = ["Age", "Fare", "Pclass", "SibSp", "Parch"]
cat_cols = ["Sex", "Embarked"]
```

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(drop="if_binary", handle_unknown="ignore", sparse_output=False)),
])
```

```python
preprocess = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols),
])
```

```python
X_train = preprocess.fit_transform(train[num_cols + cat_cols])
X_test = preprocess.transform(test[num_cols + cat_cols])

print(X_train.shape, X_test.shape)
print(preprocess.get_feature_names_out())
```

```python
df_train = pd.DataFrame(X_train, columns=preprocess.get_feature_names_out())
df_train.head()
```

```python
print(df_train["num__Age"].mean())
df_train["num__Age"].head()
```

```python
X = train[num_cols + cat_cols]
y = train["Survived"]

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier

model = Pipeline([
    ("prep", preprocess),
    ("clf", XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=3, random_state=42)),])
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(model, X, y, cv=5, scoring="accuracy")
print(scores)
print(f"{scores.mean():.3f} ± {scores.std():.3f}")
```

```python
model.fit(X, y)
predictions = model.predict(test[num_cols + cat_cols])
```

```python
submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": predictions,
})
submission.to_csv("submission.csv", index=False)
print(submission.shape)
submission.head(15)
```

```python
def add_features(df):
    df = df.copy()
    df["Title"] = df["Name"].str.extract(r" ([A-Za-z]+)\.", expand=False)
    df["Title"] = df["Title"].replace(
        ["Mlle", "Ms"], "Miss").replace("Mme", "Mrs")
    rares = ["Dr", "Rev", "Col", "Major", "Capt", "Sir", "Lady",
             "Countess", "Don", "Dona", "Jonkheer"]
    df["Title"] = df["Title"].replace(rares, "Rare")
    df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
    df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
    return df

train2 = add_features(train)
test2 = add_features(test)
```

```python
num_cols = ["Age", "Fare", "Pclass", "SibSp", "Parch", "FamilySize", "IsAlone"]
cat_cols = ["Sex", "Embarked", "Title"]

preprocess = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols),
])

model = Pipeline([
    ("prep", preprocess),
    ("clf", XGBClassifier(n_estimators=300, learning_rate=0.05,
                          max_depth=3, random_state=42)),
])

X2 = train2[num_cols + cat_cols]
y = train2["Survived"]

scores = cross_val_score(model, X2, y, cv=5, scoring="accuracy")
print(f"{scores.mean():.3f} ± {scores.std():.3f}")
```

```python
model.fit(X2, y)
predictions = model.predict(test2[num_cols + cat_cols])

submission2 = pd.DataFrame({
    "PassengerId": test2["PassengerId"],
    "Survived": predictions,
})
submission2.to_csv("submission2.csv", index=False)
print(submission2.shape)
submission2.head(15)
```
